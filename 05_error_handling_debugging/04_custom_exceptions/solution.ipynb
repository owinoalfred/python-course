```python
class CustomExc(Exception): pass
def solve_h4(fn):
    try:
        fn()
    except CustomExc as e:
        return True
    return False
```

# Robotics challenge solution — ROBO-X Challenge: Custom Exceptions Controller

**Explanation**

Robotics challenge solution for Custom Exceptions in patient-care robot monitoring.

```python
class TelemetryError(Exception): pass
def process_telemetry(data):
    if not data:
        raise TelemetryError('Empty telemetry')
    return {'status': 'OK'}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
try: process_telemetry([])
except TelemetryError: pass
```
