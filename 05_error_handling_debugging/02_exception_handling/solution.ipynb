## Solution 1 — Exception Handling Medium Exercise 1

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** try, except, finally

**Explanation and reasoning**

This solution demonstrates Exception Handling by iterating and transforming inputs.

```python
def solve_m1(a, b):
    try:
        res = a / b
    except ZeroDivisionError:
        return 'DIV_ZERO'
    else:
        return res
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Edge cases and failure modes**

Handles empty lists and None values safely.

**Alternative approaches**

Could also be implemented using generator expressions.

**Tests — run the cell to verify the reference solution**

```python
assert solve_m1(10, 2) == 5.0 and solve_m1(1, 0) == 'DIV_ZERO'
```

## Solution 9 — Exception Handling Hard Exercise 4

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** try, except, finally, else

**Explanation and reasoning**

Advanced solution applying Exception Handling with safety assertions.

```python
def solve_h4(tasks):
    results = []
    for task in tasks:
        try:
            results.append((True, task()))
        except Exception as e:
            results.append((False, str(e)))
    return results
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h4([lambda: 1])[0] == (True, 1)
```

## Solution 10 — Exception Handling Hard Exercise 5

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** try, except, finally, else

**Explanation and reasoning**

Advanced solution applying Exception Handling with safety assertions.

```python
def solve_h5(actions, cleanup):
    log = []
    try:
        for a in actions:
            a()
            log.append('ACTION')
    finally:
        cleanup()
        log.append('CLEANUP')
    return log
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h5([lambda: None], lambda: None) == ['ACTION', 'CLEANUP']
```

# Robotics challenge solution — ROBO-X Challenge: Exception Handling Controller

**Explanation**

Robotics challenge solution for Exception Handling in patient-care robot monitoring.

```python
def process_telemetry(data):
    status = 'FAIL'
    try:
        val = data[0]
        status = 'OK'
    except Exception:
        status = 'ERROR'
    finally:
        pass
    return {'status': status}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1])['status'] == 'OK'
```
