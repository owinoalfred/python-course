## 20. Engineering Example

Full engineering case study in patient-care robot monitoring:

```python
def execute_concurrent_tasks(task_list):
    errors = []
    for task in task_list:
        try:
            task()
        except Exception as e:
            errors.append(e)
    if errors:
        raise ExceptionGroup('Batch task failures', errors)
```
