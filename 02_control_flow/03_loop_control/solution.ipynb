## Solution 1 — Find the first negative reading

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** break, for, search

**Explanation and reasoning**

Returning from inside the loop stops the search at the first match, which is both the correct answer and the cheap one. The trailing return handles exhaustion without a flag variable, which would have to be initialised and then tested.

```python
def first_negative(values):
    for index, value in enumerate(values):
        if value < 0:
            return index
    return None
```

**Complexity**

Time O(n) worst case; space O(1).

**Edge cases and failure modes**

A value of exactly zero is not negative, so it does not match.

**Alternative approaches**

next((i for i, v in enumerate(values) if v < 0), None) is the generator form.

**Tests — run the cell to verify the reference solution**

```python
assert first_negative([1.0, -0.5, -2.0]) == 1
assert first_negative([]) is None
assert first_negative([0.0, 1.0]) is None
```

## Solution 2 — Sum only the usable readings

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** continue, for, filtering

**Explanation and reasoning**

A single continue covers every skip reason, which keeps the accumulation on one unindented line - the reason to prefer continue over wrapping the rest of the body in an else. Rejecting bool keeps True out of the total.

```python
def sum_usable(values) -> float:
    total = 0.0
    for value in values:
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            continue
        total += value
    return total
```

**Complexity**

Time O(n); space O(1).

**Edge cases and failure modes**

A list of only junk returns 0.0 rather than raising.

**Alternative approaches**

A generator with sum() is shorter but evaluates the same test per item.

**Tests — run the cell to verify the reference solution**

```python
assert sum_usable([1.0, None, 2.0, 'x']) == 3.0
assert sum_usable([]) == 0.0
assert sum_usable([True, 2.0]) == 2.0
```

## Solution 3 — Find the first multiple of n

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** for/else, modulo, search

**Explanation and reasoning**

The else clause states the not-found case directly, so there is no flag to initialise and no way for the two paths to disagree. Validating n before the loop avoids a ZeroDivisionError from deep inside the body.

```python
def first_multiple(values, n):
    if n == 0:
        raise ValueError('cannot divide by zero')
    for value in values:
        if value % n == 0:
            return value
    else:
        return None
```

**Complexity**

Time O(n); space O(1).

**Edge cases and failure modes**

A negative multiple is a valid match because Python's modulo handles signs.

**Alternative approaches**

next((v for v in values if v % n == 0), None) is equivalent.

**Tests — run the cell to verify the reference solution**

```python
assert first_multiple([3, 7, 10], 5) == 10
assert first_multiple([3, 7], 5) is None
```

## Solution 4 — Stop scanning at a sentinel

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** break, sentinel, search

**Explanation and reasoning**

Testing before appending is what keeps the sentinel out of the result, which is the off-by-one this exercise is really about. A sentinel that never appears simply leaves the loop to finish normally, so no special case is needed.

```python
def scan_until(values, sentinel) -> list:
    collected: list = []
    for value in values:
        if value == sentinel:
            break
        collected.append(value)
    return collected
```

**Complexity**

Time O(n); space O(n) for the result.

**Edge cases and failure modes**

A sentinel in first position returns an empty list.

**Alternative approaches**

index() plus a slice is shorter but builds a new list.

**Tests — run the cell to verify the reference solution**

```python
assert scan_until([1, 2, 9, 3], 9) == [1, 2]
assert scan_until([1, 2], 9) == [1, 2]
assert scan_until([], 9) == []
```

## Solution 5 — Detect duplicates with for/else

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** for/else, membership, sets

**Explanation and reasoning**

Checking membership before inserting is the whole algorithm: the first value that is already present is a duplicate, and returning immediately means the loop never examines the rest of the data.

```python
def has_duplicates(values) -> bool:
    seen: set = set()
    for value in values:
        if value in seen:
            return True
        seen.add(value)
    return False
```

**Complexity**

Time O(n) expected; space O(n).

**Edge cases and failure modes**

Strings and tuples are compared by value, so equal contents collide as expected.

**Alternative approaches**

len(set(values)) != len(values) is the one-liner this exercise forbids.

**Tests — run the cell to verify the reference solution**

```python
assert has_duplicates([1, 2, 1]) is True
assert has_duplicates([1, 2, 3]) is False
assert has_duplicates([]) is False
```

## Solution 6 — Locate a target in a grid

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** nested loops, return, enumeration

**Explanation and reasoning**

Returning from inside the inner loop leaves both loops at once, which is what a flag variable inside two levels of nesting cannot do cleanly. Row major order gives the first match, matching the specification.

```python
def find_target(grid, target):
    for row_index, row in enumerate(grid):
        for col_index, cell in enumerate(row):
            if cell == target:
                return row_index, col_index
    return None
```

**Complexity**

Time O(rows * cols) worst case; space O(1).

**Edge cases and failure modes**

A target of 0 is found rather than skipped, because the test is equality.

**Alternative approaches**

Flattening the grid with an index is faster but loses the coordinates.

**Tests — run the cell to verify the reference solution**

```python
assert find_target(((1, 2), (3, 4)), 3) == (1, 0)
assert find_target(((1, 2), (3, 4)), 9) is None
assert find_target((), 1) is None
```

## Solution 7 — Take readings until the budget is spent

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** break, accumulator, budget

**Explanation and reasoning**

Testing the cost against the remaining budget *before* consuming it is what guarantees the budget is never overspent. Deducting after the append keeps the two variables in step, and break ends the scan at the first item that does not fit.

```python
def spend_budget(readings, budget):
    if budget < 0:
        raise ValueError('budget must not be negative')
    consumed: list = []
    remaining = budget
    for cost in readings:
        if cost > remaining:
            break
        consumed.append(cost)
        remaining -= cost
    return consumed, remaining
```

**Complexity**

Time O(n); space O(n) for the consumed list.

**Edge cases and failure modes**

A budget of exactly one reading's cost consumes it and leaves zero.

**Alternative approaches**

A running sum with a break gives the same result.

**Tests — run the cell to verify the reference solution**

```python
assert spend_budget([3, 5, 2], 6) == ([3], 3)
assert spend_budget([1, 1], 0) == ([], 0)
assert spend_budget([2, 2], 4) == ([2, 2], 0)
```

## Solution 8 — Group consecutive equal readings

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** state, loops, grouping

**Explanation and reasoning**

The pending group is flushed when the value changes and once more after the loop, and that final flush is the detail this exercise is really about - without it the last run is silently lost.

```python
def group_runs(values) -> list[tuple]:
    groups: list[tuple] = []
    if not values:
        return groups
    current = values[0]
    count = 1
    for value in values[1:]:
        if value == current:
            count += 1
        else:
            groups.append((current, count))
            current, count = value, 1
    groups.append((current, count))
    return groups
```

**Complexity**

Time O(n); space O(k) for k runs.

**Edge cases and failure modes**

A single value produces one pair with count 1.

**Alternative approaches**

itertools.groupby produces the same structure in one line.

**Tests — run the cell to verify the reference solution**

```python
assert group_runs([1, 1, 2, 2, 2, 1]) == [(1, 2), (2, 3), (1, 1)]
assert group_runs([]) == []
assert group_runs([5]) == [(5, 1)]
```

## Solution 9 — Scan a telemetry batch for the first fault

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** continue, break, validation

**Explanation and reasoning**

Returning inside the loop means the first fault wins, whether it is a malformed entry or an overheating reading, so the report names the earliest problem rather than a later one. Testing shape before value is what makes the two fault kinds distinguishable.

```python
def first_fault(batch, max_temp):
    for index, sample in enumerate(batch):
        if not isinstance(sample, dict):
            return 'malformed', index
        temp = sample.get('t')
        if temp is not None and temp > max_temp:
            return 'overheating', index
    return None
```

**Complexity**

Time O(n); space O(1).

**Edge cases and failure modes**

A sample with no temperature key is not overheating and is skipped.

**Alternative approaches**

A validation pass followed by a search pass is clearer but slower.

**Tests — run the cell to verify the reference solution**

```python
assert first_fault([{'t': 20}, {'t': 80}], 60) == ('overheating', 1)
assert first_fault([{'t': 20}], 60) is None
assert first_fault(['x'], 60) == ('malformed', 0)
```

## Solution 10 — Drive until the battery is too low

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** break, simulator API, safety

**Explanation and reasoning**

The condition is checked before each move, so the loop stops at the first battery level at or below the limit and the battery never reaches zero. A while loop is genuinely the right choice here because termination depends on simulator state rather than on a sequence length.

```python
from shared.robo_x_sim import SimulatedRobot  # noqa: E402


def drive_until_low(robot, limit_pct: float = 20.0, step_m: float = 1.0) -> dict:
    if limit_pct <= 0:
        raise ValueError('limit_pct must be positive')
    legs = 0
    while robot.status()['battery_pct'] > limit_pct:
        robot.move(distance_m=step_m, speed_mps=1.0)
        legs += 1
    battery = robot.status()['battery_pct']
    return {'legs': legs, 'battery_pct': round(battery, 2)}
```

**Complexity**

Time O(legs); space O(1).

**Edge cases and failure modes**

A robot already below the limit completes the loop zero times.

**Alternative approaches**

move() raises on insufficient energy, which would surface as an error rather than a limit.

**Tests — run the cell to verify the reference solution**

```python
robot = SimulatedRobot(name='rover-01', battery_wh=48.0)
out = drive_until_low(robot, 20.0)
assert out['battery_pct'] <= 20.0
assert out['legs'] > 0
```

# Robotics challenge solution — ROBO-X Challenge: Mission Fault Scanner

**Explanation**

A failed read is recorded and skipped so one dead sensor does not abort the batch, while an overheating reading stops the scan immediately. The loop else reports that the whole batch was read without a fault, which is the outcome a supervisor most needs to distinguish.

```python
import sys
from pathlib import Path

_here = Path.cwd()
_root = next((p for p in [_here, *_here.parents] if (p / 'shared').is_dir()), None)
if _root is not None:
    sys.path.insert(0, str(_root))

from shared.robo_x_sim import SimulatedRobot, SensorError  # noqa: E402

MAX_TEMPERATURE_C = 60.0
CHANNELS = ('distance', 'temperature', 'battery_voltage')


def scan_batch(robot, channels=CHANNELS) -> dict:
    """Read each channel, stopping at the first genuine fault."""
    readings: dict = {}
    for channel in channels:
        try:
            value = robot.read_sensor(channel)
        except SensorError:
            readings[channel] = None
            continue
        readings[channel] = value
        if channel == 'temperature' and value > MAX_TEMPERATURE_C:
            return {'fault': 'overheating', 'readings': readings}
    else:
        return {'fault': None, 'readings': readings}
    return {'fault': 'unreachable', 'readings': readings}
```

**Complexity**

Time O(c) for c channels; space O(c).

**Notes and trade-offs**

An unknown channel name is recorded as None and does not stop the scan.

```python
robot = SimulatedRobot(name='rover-01', battery_wh=48.0)
out = scan_batch(robot, ['distance', 'temperature'])
assert out['fault'] is None
assert out['readings']['distance'] is not None
```
