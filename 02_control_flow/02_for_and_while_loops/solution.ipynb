## Solution 9 — for Loops and while Loops Hard Task 4

**Explanation and reasoning**

Advanced reference solution for for Loops and while Loops Hard Task 4. Demonstrates robust application of for Loops and while Loops.

```python
def solve_h4(stream, target):
    count = 0
    idx = 0
    while idx < len(stream):
        if stream[idx] == target:
            count += 1
        idx += 1
    return count
```

## Solution 10 — for Loops and while Loops Hard Task 5

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** for loop, while loop, Iteration

**Explanation and reasoning**

Advanced reference solution for for Loops and while Loops Hard Task 5. Demonstrates robust application of for Loops and while Loops.

```python
def solve_h5(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

# Robotics challenge solution — ROBO-X Challenge: for Loops and while Loops Controller

**Explanation**

Robotics challenge solution for for Loops and while Loops in agricultural field robot.

```python
def process_telemetry(data):
    total_dist = 0.0
    for d in data:
        total_dist += d
    return {'total_distance': total_dist}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1.0, 2.5, 0.5])['total_distance'] == 4.0
```
