## Solution 1 — Find a value's index with enumerate

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** enumerate, search, search

**Explanation and reasoning**

enumerate supplies the index from the same iteration as the value, so the two cannot drift apart - the failure mode a hand-maintained counter has once a value is skipped.

```python
def index_of(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index
    return None
```

**Complexity**

Time O(n); space O(1).

**Edge cases and failure modes**

An empty sequence returns None without an error.

**Alternative approaches**

values.index(target) is the built-in, but it raises rather than returning None.

**Tests — run the cell to verify the reference solution**

```python
assert index_of(('a', 'b', 'a'), 'a') == 0
assert index_of((), 'a') is None
```

## Solution 2 — Take every other item with a stepped range

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** range, slicing, stepping

**Explanation and reasoning**

range(0, len(values), 2) produces exactly the even indices, and the exclusive stop means the last index is only included when it really is at an even position - which is the property that makes the step reliable.

```python
def every_other(values) -> list:
    return [values[i] for i in range(0, len(values), 2)]
```

**Complexity**

Time O(n); space O(n/2).

**Edge cases and failure modes**

A one-item sequence returns that item; an empty sequence returns [].

**Alternative approaches**

values[::2] is the same result in one character pair.

**Tests — run the cell to verify the reference solution**

```python
assert every_other((1, 2, 3, 4, 5)) == [1, 3, 5]
assert every_other(()) == []
assert every_other((9,)) == [9]
```

## Solution 3 — Pair commands with battery readings

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** zip, strict, validation

**Explanation and reasoning**

zip's own strict mode is the check, so there is nothing to write. The value of the exercise is noticing that the silent default is a truncation that loses a command's reading without any signal.

```python
def pair_commands(commands, readings, strict: bool = True) -> list[tuple]:
    return list(zip(commands, readings, strict=strict))
```

**Complexity**

Time O(min(len(a), len(b))); space proportional to the result.

**Edge cases and failure modes**

With strict False a mismatch truncates silently, which is the behaviour the exercise warns about.

**Alternative approaches**

A length check before zipping gives a clearer error message but races on generators.

**Tests — run the cell to verify the reference solution**

```python
assert pair_commands(('go',), (98.0,)) == [('go', 98.0)]
import pytest
try:
    pair_commands(('a', 'b'), (1,))
except ValueError:
    pass
```

## Solution 4 — Build a countdown with a negative step

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** range, stepping, boundaries

**Explanation and reasoning**

The stop of 0 combined with a negative step is what makes 1 the final value: range stops before it would reach the stop, so 0 is never produced. A stop of 1 would drop the last element.

```python
def countdown(n: int) -> list[int]:
    return list(range(n, 0, -1))
```

**Complexity**

Time O(n); space O(n).

**Edge cases and failure modes**

countdown(0) gives an empty list, and so does a negative n.

**Alternative approaches**

range(n - 1, -1, -1) gives a 0-based countdown from n down to 0.

**Tests — run the cell to verify the reference solution**

```python
assert countdown(3) == [3, 2, 1]
assert countdown(0) == []
assert countdown(1) == [1]
```

## Solution 5 — Zip two sequences with a fill value

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** zip, itertools, padding

**Explanation and reasoning**

zip_longest is the explicit counterpart to zip: instead of stopping at the shortest input it pads it. Using it makes the padding decision visible in the call rather than hiding it behind a manual loop.

```python
from itertools import zip_longest


def zip_pad(a, b, fill=None) -> list[tuple]:
    return list(zip_longest(a, b, fillvalue=fill))
```

**Complexity**

Time O(max(len(a), len(b))).

**Edge cases and failure modes**

Two empty sequences give an empty list.

**Alternative approaches**

Pad the shorter side by hand, then use plain zip.

**Tests — run the cell to verify the reference solution**

```python
assert zip_pad((1, 2, 3), ('a',)) == [(1, 'a'), (2, None), (3, None)]
assert zip_pad((), ()) == []
```

## Solution 6 — Run-length encode a sequence

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** enumerate, grouping, itertools

**Explanation and reasoning**

groupby yields runs, not individual items, which is exactly the abstraction this exercise is about. Counting each group with a small generator sum keeps the whole thing a single comprehension.

```python
from itertools import groupby


def run_length(values) -> list[tuple]:
    return [(value, sum(1 for _ in group)) for value, group in groupby(values)]
```

**Complexity**

Time O(n); space O(r) for r runs.

**Edge cases and failure modes**

Empty input gives an empty list without special handling.

**Alternative approaches**

Tracking the previous value in a loop is the manual equivalent.

**Tests — run the cell to verify the reference solution**

```python
assert run_length(('a', 'a', 'b', 'b', 'b')) == [('a', 2), ('b', 3)]
assert run_length(()) == []
```

## Solution 7 — Transpose a table with zip

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** zip, unpacking, transformations

**Explanation and reasoning**

The star unpacking is the whole trick: zip(*rows) treats each row as a separate iterable and yields one column at a time. The length check comes first because zip would otherwise silently drop the short row's extras.

```python
def transpose(rows) -> list[tuple]:
    rows = list(rows)
    if not rows:
        return []
    if len({len(row) for row in rows}) != 1:
        raise ValueError('rows must all be the same length')
    return list(zip(*rows))
```

**Complexity**

Time O(rows * cols); space O(rows * cols).

**Edge cases and failure modes**

An empty table returns an empty list rather than raising on the unpack.

**Alternative approaches**

itertools.zip_longest pads ragged rows instead of rejecting them.

**Tests — run the cell to verify the reference solution**

```python
assert transpose(((1, 2), (3, 4))) == [(1, 3), (2, 4)]
assert transpose(()) == []
try:
    transpose(((1, 2), (3,)))
except ValueError:
    pass
```

## Solution 8 — Attach generated timestamps to readings

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** zip, generators, validation

**Explanation and reasoning**

Taking the timestamp from the index before the skip is deliberate: the time base describes when the sample was taken, so a dropped reading must leave a gap rather than shift every later time. Slicing the readings to build times would have produced the wrong answer here.

```python
def timestamped(readings, period_s: float) -> list[tuple]:
    pairs: list[tuple] = []
    for index, value in enumerate(readings):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            continue
        pairs.append((round(index * period_s, 6), value))
    return pairs
```

**Complexity**

Time O(n); space O(n).

**Edge cases and failure modes**

All-junk input gives an empty list with no timestamps consumed.

**Alternative approaches**

A generator expression is the streaming form of the same idea.

**Tests — run the cell to verify the reference solution**

```python
assert timestamped((0.1, 'x', 0.3), 0.5) == [(0.0, 0.1), (1.0, 0.3)]
assert timestamped((), 0.5) == []
```

## Solution 9 — Build a mission log from a command schedule

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** zip, strict, simulator API

**Explanation and reasoning**

The battery stream is generated from the schedule, so it is exactly as long as the commands and the two can never disagree - strict is then belt and braces rather than the primary defence. enumerate with start=1 makes the step numbers match what an operator calls step 1.

```python
import sys
from pathlib import Path

_here = Path.cwd()
_root = next((p for p in [_here, *_here.parents] if (p / 'shared').is_dir()), None)
if _root is not None:
    sys.path.insert(0, str(_root))

from shared.robo_x_sim import SimulatedRobot  # noqa: E402

def mission_log(robot, schedule) -> list[dict]:
    """One dict per command, pairing the schedule with a live reading."""
    battery = (robot.status()['battery_pct'] for _ in schedule)
    return [
        {'step': step, 'command': command, 'battery_pct': round(pct, 2)}
        for step, (command, pct) in enumerate(zip(schedule, battery, strict=True), 1)
    ]
```

**Complexity**

Time O(len(schedule)); space O(len(schedule)).

**Edge cases and failure modes**

An empty schedule gives an empty log and reads the battery zero times.

**Alternative approaches**

Reading the battery once and reusing it is faster but records a fiction.

**Tests — run the cell to verify the reference solution**

```python
robot = SimulatedRobot(name='rover-01', battery_wh=48.0)
log = mission_log(robot, ('start', 'stop'))
assert [row['step'] for row in log] == [1, 2]
assert mission_log(robot, ()) == []
```

## Solution 10 — Pair attempts with backoff delays

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** range, zip, generators

**Explanation and reasoning**

range produces the attempt numbers and the generator derives one delay per attempt, so the streams cannot disagree in length. strict=True still guards the pairing, and the generator stays lazy until list() consumes it.

```python
def retry_plan(attempts: int, base_s: float = 0.5) -> list[tuple]:
    if attempts < 0:
        raise ValueError('attempts must not be negative')
    numbers = range(1, attempts + 1)
    delays = (round(base_s * (2 ** (attempt - 1)), 6) for attempt in numbers)
    return list(zip(numbers, delays, strict=True))
```

**Complexity**

Time O(attempts); space O(attempts).

**Edge cases and failure modes**

Zero attempts gives an empty list; a negative count raises ValueError.

**Alternative approaches**

An explicit loop is clearer when the delay has a cap applied per step.

**Tests — run the cell to verify the reference solution**

```python
assert retry_plan(3) == [(1, 0.5), (2, 1.0), (3, 2.0)]
assert retry_plan(0) == []
```

# Robotics challenge solution — ROBO-X Challenge: Step-Indexed Mission Log

**Explanation**

The battery stream is generated from the readings, so strict pairing has nothing to catch and a truncated log is impossible by construction rather than by checking. The validation test happens after the pairing so a malformed reading still consumed its step, which keeps the log's step numbers aligned with the batch indices.

```python
import sys
from pathlib import Path

_here = Path.cwd()
_root = next((p for p in [_here, *_here.parents] if (p / 'shared').is_dir()), None)
if _root is not None:
    sys.path.insert(0, str(_root))

from shared.robo_x_sim import SimulatedRobot  # noqa: E402

SAMPLE_PERIOD_S = 0.5


def build_mission_log(robot, readings, period_s: float = SAMPLE_PERIOD_S) -> dict:
    """Log a fixed reading batch against a live battery trace."""
    battery = (robot.status()['battery_pct'] for _ in readings)
    entries: list[dict] = []
    for step, (value, pct) in enumerate(zip(readings, battery, strict=True), 1):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            continue
        entries.append({
            'step': step,
            't_s': round((step - 1) * period_s, 3),
            'value': value,
            'battery_pct': round(pct, 2),
        })
    return {'period_s': period_s, 'entries': entries}
```

**Complexity**

Time O(len(readings)); space O(len(readings)).

**Notes and trade-offs**

An empty batch gives an empty entries list and reads the battery zero times.

```python
robot = SimulatedRobot(name='rover-01', battery_wh=48.0)
out = build_mission_log(robot, (0.1, 0.2))
assert [e['step'] for e in out['entries']] == [1, 2]
assert build_mission_log(robot, ())['entries'] == []
```
