## Solution 7 — Regular Expressions Hard Exercise 2

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Regex, re, Pattern Matching

**Explanation and reasoning**

Advanced solution applying Regular Expressions with safety assertions.

```python
def solve_h2(text):
    import re
    pattern = r'^(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})$'
    m = re.match(pattern, text)
    return m.groupdict() if m else {}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h2('2026-10-03')['year'] == '2026'
```

# Robotics challenge solution — ROBO-X Challenge: Regular Expressions Controller

**Explanation**

Robotics challenge solution for Regular Expressions in flight-test data pipeline.

```python
def process_telemetry(data):
    import re
    raw_str = ' '.join(str(x) for x in data)
    numbers = [float(n) for n in re.findall(r'[-+]?\d*\.?\d+', raw_str)]
    return {'parsed_numbers': numbers}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry(['speed=12.5m/s'])['parsed_numbers'] == [12.5]
```
