## 12. Code Walkthrough

Detailed walkthrough of a complete Regular Expressions implementation:

```python
import re
line = 'SENSOR_1: TEMP=42.5 C'
m = re.search(r'TEMP=(\d+\.\d+)', line)
if m:
    print('Temperature:', m.group(1))
```

## 20. Engineering Example

Full engineering case study in flight-test data pipeline:

```python
import re
def parse_log_line(line):
    pattern = r'^\[(?P<level>\w+)\]\s+(?P<msg>.*)$'
    m = re.match(pattern, line)
    return m.groupdict() if m else {}
```
