## Solution 6 — Working with Paths Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** pathlib, Path, glob

**Explanation and reasoning**

Advanced solution applying Working with Paths with safety assertions.

```python
def solve_h1(path_str):
    from pathlib import Path
    p = Path(path_str)
    return {'name': p.name, 'stem': p.stem, 'suffix': p.suffix, 'parent': str(p.parent)}
```

# Robotics challenge solution — ROBO-X Challenge: Working with Paths Controller

**Explanation**

Robotics challenge solution for Working with Paths in flight-test data pipeline.

```python
def process_telemetry(data):
    from pathlib import Path
    p = Path('telemetry/readings.csv')
    return {'ext': p.suffix, 'count': len(data)}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1])['ext'] == '.csv'
```
