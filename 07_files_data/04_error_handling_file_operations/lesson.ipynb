Intermediate engineering application of Error Handling in File Operations:

```python
import json
def safe_load_json(filepath, default={}):
    try:
        with open(filepath) as f:
            return json.load(f)
    except (FileNotFoundError, json.JSONDecodeError):
        return default
```
