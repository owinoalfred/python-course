```python
def solve_h4(d1, d2, key):
    import pandas as pd
    df1 = pd.DataFrame(d1)
    df2 = pd.DataFrame(d2)
    merged = pd.merge(df1, df2, on=key)
    return merged.to_dict(orient='records')
```

# Robotics challenge solution — ROBO-X Challenge: Core Data Libraries Controller

**Explanation**

Robotics challenge solution for Core Data Libraries in flight-test data pipeline.

```python
def process_telemetry(data):
    import pandas as pd
    df = pd.DataFrame({'sensor_val': data})
    return {'avg_val': float(df['sensor_val'].mean()) if not df.empty else 0.0}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([10.0, 20.0])['avg_val'] == 15.0
```
