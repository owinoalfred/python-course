## Solution 6 — OOP Fundamentals Hard Task 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Class, Instance, __init__, self

**Explanation and reasoning**

Advanced reference solution for OOP Fundamentals Hard Task 1. Demonstrates robust application of OOP Fundamentals.

```python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    def deposit(self, amt):
        self.balance += amt
    def withdraw(self, amt):
        if amt <= self.balance:
            self.balance -= amt
            return True
        return False
def solve_h1():
    b = BankAccount(100)
    b.deposit(50)
    ok = b.withdraw(120)
    return b.balance, ok
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and safely handles malformed or missing elements.

**Alternative approaches**

Can be implemented using declarative or object-oriented design patterns.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h1() == (30, True)
```

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Class, Instance, __init__, self

**Explanation and reasoning**

Advanced reference solution for OOP Fundamentals Hard Task 4. Demonstrates robust application of OOP Fundamentals.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add(self, item):
        self.items.append(item)
    def total(self):
        return sum(i.price for i in self.items)
def solve_h4():
    cart = ShoppingCart()
    cart.add(Item('a', 10))
    cart.add(Item('b', 20))
    return cart.total()
```
