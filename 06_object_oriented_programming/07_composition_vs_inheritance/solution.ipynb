## Solution 6 — Composition vs Inheritance Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Composition, Inheritance, HAS-A, IS-A

**Explanation and reasoning**

Advanced solution applying Composition vs Inheritance with safety assertions.

```python
class Sensor:
    def __init__(self, name): self.name = name
    def read(self): return 42.0
class Robot:
    def __init__(self, names):
        self.sensors = [Sensor(n) for n in names]
    def poll(self): return {s.name: s.read() for s in self.sensors}
def solve_h1(): r = Robot(['s1', 's2']); return r.poll()
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h1() == {'s1': 42.0, 's2': 42.0}
```

# Robotics challenge solution — ROBO-X Challenge: Composition vs Inheritance Controller

**Explanation**

Robotics challenge solution for Composition vs Inheritance in modular robotic arm.

```python
class Sensor:
    def read(self): return 10.0
class Robot:
    def __init__(self):
        self.s = Sensor()
    def status(self): return {'sensor_val': self.s.read()}
def process_telemetry(data):
    r = Robot()
    return r.status()
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([])['sensor_val'] == 10.0
```
