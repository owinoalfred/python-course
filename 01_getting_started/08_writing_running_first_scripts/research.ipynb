## Method

This is **not** a search-and-summarise assignment. You must produce real measurements on your machine, then explain what they mean.

1. State a question that can be settled by measurement.
2. Write a falsifiable hypothesis — one that *could* be wrong.
3. Design an experiment that isolates one variable.
4. Run it, collect data, and keep the raw numbers.
5. Analyse: what would each possible result have meant?
6. Record the result, including results that surprised you.
7. Interpret the mechanism, not just the numbers.
8. Conclude, and state what you would measure next.

> **Measure, do not assume**
>
> Every claim in RESULT, INTERPRETATION and CONCLUSION must be traceable to a number you produced in this notebook. If you cannot cite the measurement, it is speculation — label it as such.
