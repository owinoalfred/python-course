## Project brief

Build a small, fully tested command-line tool for controlling a simulated robot fleet. It must be safe to import, exit with meaningful codes, and keep its data and diagnostics on separate streams.

**Scenario**

A deployment script calls `roboctl status` and branches on the result. The tool must be importable by the test suite without starting a robot, and its output must be machine-readable so the script can parse it.

**Why this project exists**

This is the shape of every robotics service that will be written in this course. Getting the process contract right once means the pattern is already in place when the code grows.

## Requirements

- Put the real logic in a function that takes argv and both streams.
- Support status, stop and help, returning 0, 1 and 2 as appropriate.
- Print one parseable line per robot to stdout.
- Send usage and unknown-command messages to stderr only.
- Add an `if __name__ == '__main__':` guard that exits once.

**Constraints**

- Standard library only.
- No function may read sys.argv or call print without a stream.
- Importing the module must start no robot and print nothing.

**Deliverables**

- `roboctl.py` with the dispatch logic and a thin main.
- `test_roboctl.py` with at least twelve assertions.
- A README section documenting every command and every exit code.

## Implementation steps

Work through these in order. Do not start at step 4 — each step is designed to be runnable and testable before the next begins.

1. Write a `render_status(robot)` function that returns a line and prints nothing.
2. Write `run(argv, out, errors)` returning an exit code.
3. Add the main guard and confirm `python roboctl.py status` works from a shell.
4. Write tests that import the module and assert importing printed nothing.
5. Add a test that runs the file as a subprocess and checks the return code.
