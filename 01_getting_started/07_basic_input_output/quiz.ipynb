**Q1 — Answer: `B`**

print looks up sys.stdout at call time and writes to whatever object it currently references. Because that attribute can be rebound, redirecting it changes where every later print lands.

*Reference:* lesson.ipynb - Formal Theory

**Q2 — Answer: `A`**

input always returns a string regardless of what was typed. The value must be converted explicitly before arithmetic, which is the source of the classic can-only-concatenate-str-to-str error.

*Reference:* lesson.ipynb - Common Mistakes

**Q3 — Answer: `C`**

The redirect captured everything on stdout, so the diagnostics became part of the data. The fix is the CLI contract: stdout carries results, stderr carries messages meant for a human.

*Reference:* lesson.ipynb - Common Mistakes

**Q4 — Answer: `D`**

readline returns the empty string only at end of input; a blank line is returned as a newline. Without the check, a closed pipe turns a read loop into an infinite loop.

*Reference:* lesson.ipynb - Code Walkthrough

**Q5 — Answer: `A`**

It rebinds sys.stdout for the block and restores it afterwards, even if the block raises. That is exactly what a test needs to assert on printed output without a terminal.

*Reference:* lesson.ipynb - Advanced Examples

**Q6 — Answer: `C`**

The alignment character comes first, then the minimum width, then the precision. '>8f' omits the precision and '8.1f' omits the alignment, so only '>8.1f' produces right alignment with a single decimal place.

*Reference:* lesson.ipynb - Conceptual Explanation

**Q7 — Answer: `B`**

Repeated += copies the accumulated string each time, making the loop quadratic in the number of lines. A list with append is amortised linear and a single join produces the same result.

*Reference:* lesson.ipynb - Pythonic Approaches

**Q8 — Answer: `C`**

A stream parameter is the seam that makes console code testable: a test supplies a StringIO and asserts on the result. Performance is a side effect of the pattern, not its purpose.

*Reference:* exercises.ipynb - Exercise 10

**Q9 — Answer: `B`**

When stdout is not a terminal Python block-buffers it, so a service can sit in a sleep loop with nothing written yet. Passing flush=True, or running with python -u, restores the interactive behaviour.

*Reference:* lesson.ipynb - Debugging Techniques

**Q10 — Answer: `A`**

The supervisor parses stdout, so any warning written there corrupts the data stream. stderr is not captured by the standard redirect, which makes it the correct home for diagnostics in a service.

*Reference:* solution.ipynb - Solution 8
