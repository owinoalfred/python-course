## Solution 1 — Comprehensions Medium Exercise 1

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** Comprehensions, Generator Expressions

**Explanation and reasoning**

This solution demonstrates Comprehensions by iterating and transforming inputs.

```python
def solve_m1(n):
    return [x**2 for x in range(n)]
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Edge cases and failure modes**

Handles empty lists and None values safely.

**Alternative approaches**

Could also be implemented using generator expressions.

**Tests — run the cell to verify the reference solution**

```python
assert solve_m1(4) == [0, 1, 4, 9]
```

## Solution 6 — Comprehensions Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Comprehensions, Generator Expressions

**Explanation and reasoning**

Advanced solution applying Comprehensions with safety assertions.

```python
def solve_h1(n):
    return [x for x in range(2, n) if all(x % d != 0 for d in range(2, int(x**0.5) + 1))]
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h1(10) == [2, 3, 5, 7]
```

# Robotics challenge solution — ROBO-X Challenge: Comprehensions Controller

**Explanation**

Robotics challenge solution for Comprehensions in cleanroom material handler.

```python
def process_telemetry(data):
    scaled = [x * 100.0 for x in data if x >= 0]
    return {'scaled': scaled}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([0.1, -0.5, 0.2])['scaled'] == [10.0, 20.0]
```
