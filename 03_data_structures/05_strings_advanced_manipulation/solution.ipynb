## Solution 9 — Strings and Advanced String Manipulation Hard Exercise 4

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Strings, Immutability, Log Parsing

**Explanation and reasoning**

Advanced solution applying Strings and Advanced String Manipulation with safety assertions.

```python
def solve_h4(s):
    res = []
    for char in s:
        if char.isupper():
            res.append(char.lower())
        else:
            res.append(char.upper())
    return ''.join(res)
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h4('PyThOn') == 'pYtHoN'
```

## Solution 10 — Strings and Advanced String Manipulation Hard Exercise 5

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Strings, Immutability, Log Parsing

**Explanation and reasoning**

Advanced solution applying Strings and Advanced String Manipulation with safety assertions.

```python
def solve_h5(s):
    cleaned = ''.join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]
```

# Robotics challenge solution — ROBO-X Challenge: Strings and Advanced String Manipulation Controller

**Explanation**

Robotics challenge solution for Strings and Advanced String Manipulation in port crane telemetry.

```python
def process_telemetry(data):
    msg = str(data[0]) if data else ''
    return {'clean_msg': msg.upper().strip()}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry(['  status_ok '])['clean_msg'] == 'STATUS_OK'
```
