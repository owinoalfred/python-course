## Solution 6 — Advanced Function Concepts Hard Task 1

**Explanation and reasoning**

Advanced reference solution for Advanced Function Concepts Hard Task 1. Demonstrates robust application of Advanced Function Concepts.

```python
def solve_h1(nested):
    total = 0
    for x in nested:
        if isinstance(x, list):
            total += solve_h1(x)
        else:
            total += x
    return total
```

## Solution 7 — Advanced Function Concepts Hard Task 2

**Explanation and reasoning**

Advanced reference solution for Advanced Function Concepts Hard Task 2. Demonstrates robust application of Advanced Function Concepts.

```python
def solve_h2(fn, times):
    def repeated(x):
        res = x
        for _ in range(times):
            res = fn(res)
        return res
    return repeated
```

## Solution 10 — Advanced Function Concepts Hard Task 5

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Higher-Order Functions, Recursion

**Explanation and reasoning**

Advanced reference solution for Advanced Function Concepts Hard Task 5. Demonstrates robust application of Advanced Function Concepts.

```python
def solve_h5(arr):
    if len(arr) <= 1: return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return solve_h5(left) + middle + solve_h5(right)
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and safely handles malformed or missing elements.

**Alternative approaches**

Can be implemented using declarative or object-oriented design patterns.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h5([3, 1, 2]) == [1, 2, 3]
```

# Robotics challenge solution — ROBO-X Challenge: Advanced Function Concepts Controller

**Explanation**

Robotics challenge solution for Advanced Function Concepts in manufacturing cell controller.

```python
def process_telemetry(data):
    def recursive_sum(lst):
        if not lst: return 0.0
        return float(lst[0]) + recursive_sum(lst[1:])
    return {'sum': recursive_sum(data)}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1.0, 2.0])['sum'] == 3.0
```
