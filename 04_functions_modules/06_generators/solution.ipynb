## Solution 1 — Generators Medium Exercise 1

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** Generators, yield, Lazy Evaluation

**Explanation and reasoning**

This solution demonstrates Generators by iterating and transforming inputs.

```python
def solve_m1(n):
    for i in range(n):
        yield i
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Edge cases and failure modes**

Handles empty lists and None values safely.

**Alternative approaches**

Could also be implemented using generator expressions.

**Tests — run the cell to verify the reference solution**

```python
assert list(solve_m1(3)) == [0, 1, 2]
```

## Solution 6 — Generators Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Generators, yield, Lazy Evaluation

**Explanation and reasoning**

Advanced solution applying Generators with safety assertions.

```python
def solve_h1(stream, window_size):
    from collections import deque
    buf = deque(maxlen=window_size)
    for x in stream:
        buf.append(x)
        if len(buf) == window_size:
            yield list(buf)
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert list(solve_h1([1, 2, 3, 4], 2)) == [[1, 2], [2, 3], [3, 4]]
```

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Generators, yield, Lazy Evaluation

**Explanation and reasoning**

Advanced solution applying Generators with safety assertions.

```python
def solve_h5(stream):
    prev = None
    for x in stream:
        if x != prev:
            yield x
            prev = x
```

**Alternative approaches**

# Robotics challenge solution — ROBO-X Challenge: Generators Controller

**Explanation**

Robotics challenge solution for Generators in subsea ROV.

```python
def process_telemetry(data):
    def gen():
        for x in data:
            yield x * 2.0
    return {'stream_sample': list(gen())}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1.0, 2.0])['stream_sample'] == [2.0, 4.0]
```
