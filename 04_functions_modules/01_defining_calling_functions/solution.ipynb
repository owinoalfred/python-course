## Solution 8 — Defining and Calling Functions Hard Exercise 3

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Functions, def, Docstrings

**Explanation and reasoning**

Advanced solution applying Defining and Calling Functions with safety assertions.

```python
def solve_h3(funcs, val):
    res = val
    for f in funcs:
        res = f(res)
    return res
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h3([lambda x: x+1, lambda x: x*2], 3) == 8
```
