# Robotics challenge solution — ROBO-X Challenge: Parameters and Arguments Controller

**Explanation**

Robotics challenge solution for Parameters and Arguments in manufacturing cell controller.

```python
def process_telemetry(data, *args, **kwargs):
    return {'data_len': len(data), 'extra': len(args) + len(kwargs)}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1], 2, x=3)['extra'] == 2
```
