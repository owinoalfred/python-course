## Solution 6 — Modules and Packages Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Modules, Packages, __init__.py

**Explanation and reasoning**

Advanced solution applying Modules and Packages with safety assertions.

```python
def solve_h1(pkg_path):
    from pathlib import Path
    p = Path(pkg_path)
    return [f.stem for f in p.glob('*.py') if f.name != '__init__.py']
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

```python
assert isinstance(solve_h1('tools'), list)
```
