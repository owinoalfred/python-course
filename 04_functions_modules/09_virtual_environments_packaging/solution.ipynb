## Solution 1 — Virtual Environments and Package Management Medium Exercise 1

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** venv, pip, pyproject.toml

**Explanation and reasoning**

This solution demonstrates Virtual Environments and Package Management by iterating and transforming inputs.

```python
def solve_m1(req_str):
    return [line.strip() for line in req_str.splitlines() if line.strip() and not line.startswith('#')]
```

**Complexity**

## Solution 6 — Virtual Environments and Package Management Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** venv, pip, pyproject.toml

**Explanation and reasoning**

Advanced solution applying Virtual Environments and Package Management with safety assertions.

```python
def solve_h1(pyproject_content):
    lines = pyproject_content.splitlines()
    in_deps = False
    deps = []
    for line in lines:
        if 'dependencies =' in line:
            in_deps = True
            continue
        if in_deps:
            if line.strip().endswith(']'): break
            clean = line.strip().strip('",')
            if clean: deps.append(clean)
    return deps
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h1('dependencies = [
"numpy>=1.20",
]') == ['numpy>=1.20']
```
