## Solution 1 — Lambda Functions and Functional Tools Medium Task 1

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** lambda, map, filter

**Explanation and reasoning**

Reference implementation for Lambda Functions and Functional Tools Medium Task 1. Demonstrates Lambda Functions and Functional Tools by validating and transforming inputs.

```python
def solve_m1(a, b):
    f = lambda x, y: x * y
    return f(a, b)
```

## Solution 10 — Lambda Functions and Functional Tools Hard Task 5

# Robotics challenge solution — ROBO-X Challenge: Lambda Functions and Functional Tools Controller

**Explanation**

Robotics challenge solution for Lambda Functions and Functional Tools in subsea ROV.

```python
def process_telemetry(data):
    filtered = list(filter(lambda x: x >= 0, data))
    return {'valid': filtered}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1.0, -2.0])['valid'] == [1.0]
```
