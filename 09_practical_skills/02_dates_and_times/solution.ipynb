## Solution 9 — Dates and Times Hard Exercise 4

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** datetime, timedelta, ISO-8601, Timezone

**Explanation and reasoning**

Advanced solution applying Dates and Times with safety assertions.

```python
def solve_h4(timestamps_iso):
    from datetime import datetime
    dts = [datetime.fromisoformat(ts) for ts in timestamps_iso]
    return [dts[i+1] - dts[i] for i in range(len(dts)-1)]
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

# Robotics challenge solution — ROBO-X Challenge: Dates and Times Controller

**Explanation**

Robotics challenge solution for Dates and Times in robotics-as-a-service platform.

```python
def process_telemetry(data):
    from datetime import datetime, timezone
    ts = datetime.now(timezone.utc).isoformat()
    return {'timestamp': ts, 'data_len': len(data)}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert 'T' in process_telemetry([1])['timestamp']
```
