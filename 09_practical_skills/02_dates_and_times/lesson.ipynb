## 20. Engineering Example

Full engineering case study in robotics-as-a-service platform:

```python
from datetime import datetime, timezone
def is_telemetry_fresh(timestamp_iso: str, max_age_s: float = 5.0) -> bool:
    try:
        dt = datetime.fromisoformat(timestamp_iso)
        now = datetime.now(timezone.utc)
        return (now - dt).total_seconds() <= max_age_s
    except Exception:
        return False
```
