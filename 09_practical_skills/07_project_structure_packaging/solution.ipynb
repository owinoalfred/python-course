Advanced reference solution for Project Structure and Packaging Hard Task 1. Demonstrates robust application of Project Structure and Packaging.

## Solution 10 — Project Structure and Packaging Hard Task 5

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** Project Structure, pyproject.toml, Packaging, Wheels

**Explanation and reasoning**

Advanced reference solution for Project Structure and Packaging Hard Task 5. Demonstrates robust application of Project Structure and Packaging.

```python
def solve_h5(raw_version):
    import re
    parts = re.findall(r'\d+', raw_version)
    return '.'.join(parts[:3]) if len(parts) >= 3 else '0.1.0'
```
