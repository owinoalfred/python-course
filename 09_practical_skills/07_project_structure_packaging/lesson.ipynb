## 20. Engineering Example

Full engineering case study in robotics-as-a-service platform:

```python
from pathlib import Path
def init_package_structure(target_dir, pkg_name):
    root = Path(target_dir)
    pkg_dir = root / 'src' / pkg_name
    pkg_dir.mkdir(parents=True, exist_ok=True)
    (pkg_dir / '__init__.py').write_text(f'"""{pkg_name} package"""\n__version__ = "0.1.0"\n')
```
