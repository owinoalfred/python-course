## Solution 6 — Unit Testing Hard Task 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** pytest, Unit Testing, Fixtures, Mocking

**Explanation and reasoning**

Advanced reference solution for Unit Testing Hard Task 1. Demonstrates robust application of Unit Testing.

```python
def solve_h1(target_fn, mock_val):
    from unittest.mock import patch
    with patch(target_fn) as mock:
        mock.return_value = mock_val
        return mock()
    return None
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and safely handles malformed or missing elements.

**Alternative approaches**

Can be implemented using declarative or object-oriented design patterns.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h1('os.getcwd', '/mock/dir') == '/mock/dir'
```
