## 20. Engineering Example

Full engineering case study in robotics-as-a-service platform:

```python
import requests
def post_robot_telemetry(api_url, payload, timeout_s=3.0):
    try:
        resp = requests.post(api_url, json=payload, timeout=timeout_s)
        resp.raise_for_status()
        return True, resp.json()
    except requests.RequestException as e:
        return False, {'error': str(e)}
```
