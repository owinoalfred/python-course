## Solution 5 — HTTP Requests and APIs Medium Exercise 5

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** requests, REST API, HTTP Methods

**Explanation and reasoning**

This solution demonstrates HTTP Requests and APIs by iterating and transforming inputs.

```python
def solve_m5(status_code):
    import requests
    resp = requests.Response()
    resp.status_code = status_code
    try:
        resp.raise_for_status()
        return True
    except requests.HTTPError:
        return False
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Edge cases and failure modes**

Handles empty lists and None values safely.

**Alternative approaches**

Could also be implemented using generator expressions.

**Tests — run the cell to verify the reference solution**

```python
assert solve_m5(200) is True and solve_m5(500) is False
```
