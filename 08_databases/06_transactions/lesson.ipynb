## 9. Basic Examples

Basic usage pattern for Transactions:

```python
import sqlite3
conn = sqlite3.connect(':memory:')
conn.execute('CREATE TABLE t (x INT)')
try:
    conn.execute('INSERT INTO t VALUES (1)')
    conn.commit()
except Exception:
    conn.rollback()
```

## 11. Advanced Examples

Advanced scenario demonstrating Transactions in complex robotics workflows:

```python
import sqlite3
conn = sqlite3.connect(':memory:')
conn.execute('CREATE TABLE t (x INT PRIMARY KEY)')
try:
    with conn:
        conn.execute('INSERT INTO t VALUES (1)')
        conn.execute('INSERT INTO t VALUES (1)') # Duplicate primary key error!
except sqlite3.IntegrityError:
    print('Transaction rolled back automatically')
```

## 12. Code Walkthrough

Detailed walkthrough of a complete Transactions implementation:

```python
import sqlite3
conn = sqlite3.connect(':memory:')
conn.execute('CREATE TABLE accounts (id INT, bal REAL)')
conn.execute('INSERT INTO accounts VALUES (1, 100.0), (2, 50.0)')
conn.commit()
try:
    conn.execute('UPDATE accounts SET bal = bal - 30 WHERE id = 1')
    conn.execute('UPDATE accounts SET bal = bal + 30 WHERE id = 2')
    conn.commit()
except Exception:
    conn.rollback()
print('Balances:', conn.execute('SELECT bal FROM accounts').fetchall())
```

## 20. Engineering Example

Full engineering case study in fleet operations centre:

```python
def transfer_funds(conn, src_id, dst_id, amount):
    try:
        with conn:
            conn.execute('UPDATE accounts SET bal = bal - ? WHERE id = ?', (amount, src_id))
            conn.execute('UPDATE accounts SET bal = bal + ? WHERE id = ?', (amount, dst_id))
        return True
    except sqlite3.Error:
        return False
```
