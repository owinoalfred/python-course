## Solution 1 — Parameterized Queries and SQL Injection Medium Exercise 1

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** SQL Injection, Parameterized Queries, Placeholders

**Explanation and reasoning**

This solution demonstrates Parameterized Queries and SQL Injection by iterating and transforming inputs.

```python
def solve_m1(val):
    import sqlite3
    conn = sqlite3.connect(':memory:')
    conn.execute('CREATE TABLE t (x TEXT);')
    conn.execute('INSERT INTO t VALUES (?);', (val,))
    return conn.execute('SELECT x FROM t WHERE x = ?;', (val,)).fetchone()[0]
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Edge cases and failure modes**

Handles empty lists and None values safely.

**Alternative approaches**

Could also be implemented using generator expressions.

**Tests — run the cell to verify the reference solution**

```python
assert solve_m1("' OR 1=1 --") == "' OR 1=1 --"
```

## Solution 7 — Parameterized Queries and SQL Injection Hard Exercise 2

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** SQL Injection, Parameterized Queries, Placeholders

**Explanation and reasoning**

Advanced solution applying Parameterized Queries and SQL Injection with safety assertions.

```python
def solve_h2(filters_dict):
    import sqlite3
    conn = sqlite3.connect(':memory:')
    conn.execute('CREATE TABLE t (a INT, b TEXT);')
    conditions = []
    params = []
    for k, v in filters_dict.items():
        conditions.append(f'{k} = ?')
        params.append(v)
    where_clause = ' AND '.join(conditions) if conditions else '1=1'
    query = f'SELECT * FROM t WHERE {where_clause};'
    return query, params
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Can use imperative filtering loops.

# Robotics challenge solution — ROBO-X Challenge: Parameterized Queries and SQL Injection Controller

**Explanation**

Robotics challenge solution for Parameterized Queries and SQL Injection in fleet operations centre.

```python
def process_telemetry(data):
    import sqlite3
    conn = sqlite3.connect(':memory:')
    conn.execute('CREATE TABLE telemetry (val REAL);')
    val = data[0] if data else 0.0
    conn.execute('INSERT INTO telemetry VALUES (?);', (val,))
    res = conn.execute('SELECT val FROM telemetry WHERE val = ?;', (val,)).fetchone()[0]
    return {'parameterized_val': res}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([42.0])['parameterized_val'] == 42.0
```
