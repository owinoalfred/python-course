## 20. Engineering Example

Full engineering case study in fleet operations centre:

```python
def build_fleet_query(min_battery, active_only=True):
    conditions = [f'battery_pct >= {min_battery}']
    if active_only:
        conditions.append("status = 'ACTIVE'")
    return f'SELECT * FROM fleet WHERE {" AND ".join(conditions)};'
```
