## Solution 6 — External Databases Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** PostgreSQL, MySQL, psycopg2, Connection Pooling

**Explanation and reasoning**

Advanced solution applying External Databases with safety assertions.

```python
def solve_h1(db_type, user, pwd, host, port, db):
    drivers = {'postgres': f'postgresql://{user}:{pwd}@{host}:{port}/{db}', 'mysql': f'mysql+mysqlconnector://{user}:{pwd}@{host}:{port}/{db}'}
    return drivers.get(db_type, 'sqlite:///:memory:')
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

```python
assert solve_h1('postgres', 'u', 'p', 'h', 5432, 'd').startswith('postgresql')
```

# Robotics challenge solution — ROBO-X Challenge: External Databases Controller

**Explanation**

Robotics challenge solution for External Databases in fleet operations centre.

```python
def process_telemetry(data):
    db_url = 'sqlite:///:memory:'
    return {'active_db_url': db_url, 'count': len(data)}
```

**Complexity**

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1])['active_db_url'] == 'sqlite:///:memory:'
```
