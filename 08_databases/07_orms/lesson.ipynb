## 9. Basic Examples

Basic usage pattern for ORMs:

```python
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.orm import declarative_base
Base = declarative_base()
class Robot(Base):
    __tablename__ = 'robots'
    id = Column(Integer, primary_key=True)
    name = Column(String)
engine = create_engine('sqlite:///:memory:')
Base.metadata.create_all(engine)
```

## 10. Intermediate Examples

Intermediate engineering application of ORMs:

```python
from sqlalchemy.orm import sessionmaker
Session = sessionmaker(bind=engine)
session = Session()
r1 = Robot(id=1, name='ROBO-1')
session.add(r1)
session.commit()
```

## 11. Advanced Examples

## 12. Code Walkthrough

Detailed walkthrough of a complete ORMs implementation:

```python
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.orm import declarative_base, sessionmaker
Base = declarative_base()
class FleetRobot(Base):
    __tablename__ = 'fleet'
    id = Column(Integer, primary_key=True)
    model = Column(String)
engine = create_engine('sqlite:///:memory:')
Base.metadata.create_all(engine)
Session = sessionmaker(bind=engine)
s = Session()
s.add(FleetRobot(id=1, model='X1'))
s.commit()
r = s.query(FleetRobot).first()
print('ORM Robot:', r.model)
```
