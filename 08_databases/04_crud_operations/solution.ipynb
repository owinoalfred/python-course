## Solution 1 — CRUD Operations Medium Exercise 1

**Difficulty:** `MEDIUM` &nbsp;|&nbsp; **Concepts:** CRUD, INSERT, SELECT

**Explanation and reasoning**

This solution demonstrates CRUD Operations by iterating and transforming inputs.

```python
def solve_m1(table, val):
    import sqlite3
    conn = sqlite3.connect(':memory:')
    conn.execute(f'CREATE TABLE {table} (x INT);')
    conn.execute(f'INSERT INTO {table} VALUES (?);', (val,))
    return conn.execute(f'SELECT x FROM {table};').fetchone()[0]
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Edge cases and failure modes**

Handles empty lists and None values safely.

**Alternative approaches**

Could also be implemented using generator expressions.

**Tests — run the cell to verify the reference solution**

```python
assert solve_m1('t', 10) == 10
```

## Solution 6 — CRUD Operations Hard Exercise 1

**Difficulty:** `HARD` &nbsp;|&nbsp; **Concepts:** CRUD, INSERT, SELECT, UPDATE, DELETE

**Explanation and reasoning**

Advanced solution applying CRUD Operations with safety assertions.

```python
def solve_h1(records):
    import sqlite3
    conn = sqlite3.connect(':memory:')
    conn.execute('CREATE TABLE t (id INT PRIMARY KEY, name TEXT);')
    for r in records:
        conn.execute('INSERT INTO t VALUES (?, ?) ON CONFLICT(id) DO UPDATE SET name=excluded.name;', r)
    return conn.execute('SELECT name FROM t WHERE id = 1;').fetchone()[0]
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(N)

**Edge cases and failure modes**

Validates input types and ignores malformed elements.

**Alternative approaches**

Can use imperative filtering loops.

**Tests — run the cell to verify the reference solution**

# Robotics challenge solution — ROBO-X Challenge: CRUD Operations Controller

**Explanation**

Robotics challenge solution for CRUD Operations in fleet operations centre.

```python
def process_telemetry(data):
    import sqlite3
    conn = sqlite3.connect(':memory:')
    conn.execute('CREATE TABLE telemetry (val REAL);')
    conn.executemany('INSERT INTO telemetry VALUES (?);', [(x,) for x in data])
    rows = conn.execute('SELECT * FROM telemetry;').fetchall()
    return {'crud_read': len(rows)}
```

**Complexity**

Time Complexity: O(N), Space Complexity: O(1)

**Notes and trade-offs**

Handles sensor noise and timeout conditions.

```python
assert process_telemetry([1.0, 2.0])['crud_read'] == 2
```
